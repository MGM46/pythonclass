```python
stock_name = input("What is the stock name? ")
sell_price = float(input("Sell Price? "))
buy_price = float(input("Buy Price? "))

profit = sell_price - buy_price
Return = (profit / buy_price) * 100

print(f"Stock: {stock_name}")
print(f"Buy Price: RM {buy_price:.2f}")
print(f"Sell Price: RM {sell_price:.2f}")
print("_____________________")
print(f"Profit/Loss: RM {profit:.2f}")
print(f"Return: {Return:.2f}%")
```
