```python
employees = [
    {"name": "Ahmad",    "monthly_salary": 2500, "company": "Petronas"},
    {"name": "Mei Ling", "monthly_salary": 5000, "company": "Maybank"},
    {"name": "Arif",     "monthly_salary": 8000, "company": "CIMB"},
]

# Loop through each employee
# Step 1: calculate annual salary (monthly × 12)
# Step 2: classify into B40 / M40 / T20
# Step 3: print the result

for employee in employees:
    annual_salary = employee["monthly_salary"]*12

    if annual_salary  > 80000:
        income_group = "T20"
    elif annual_salary >= 40000:
        income_group = "M40"
    else:
        income_group = "B40"

    print(f"{employee['name']} ({employee['company']}): {income_group}(RM {annual_salary}/year)")
```

```python
#Investment return calculator
def calculate_return(Selling_price, Buying_price):
    Profit = Selling_price - Buying_price
    Return = ((Selling_price - Buying_price)/Buying_price)*100
    print(f"Profit: RM {Profit}")
    print(f"Return: {Return:.2f}%")

calculate_return(20.56, 12.12)
```

```python
#Compound interest calculator
def FV(PV, rate, n):
    Future_Value = PV*((1+rate)**n)
    print(f"PV: {PV}")
    print(f"Rate: {rate}")
    print(f"Year: {n}")
    print(f"FV: {Future_Value:.2f}")
FV(1000,0.05,10)
```

```python
import pandas as pd
import quantstats as qs

url = "https://raw.githubusercontent.com/nextynewb/datasets/main/aapl.csv"
df = pd.read_csv(url, parse_dates=["Date"], index_col="Date")

# daily % change, drop the first NaN
returns = df["Close"].pct_change().dropna()

# annualized Sharpe — one call
sharpe = qs.stats.sharpe(returns)

print(f"Sharpe ratio: {sharpe:.2f}")
```

```python
import pandas as pd
import quantstats as qs

url = "https://raw.githubusercontent.com/nextynewb/datasets/main/aapl.csv"
df = pd.read_csv(url, parse_dates=["Date"], index_col="Date")

returns = df["Close"].pct_change().dropna()

# ── Risk-adjusted return ──
print(qs.stats.sharpe(returns))        # 1.56
print(qs.stats.sortino(returns))       # penalise downside only
print(qs.stats.calmar(returns))        # CAGR ÷ max drawdown

# ── Drawdown ──
print(qs.stats.max_drawdown(returns))  # worst peak-to-trough

# ── Growth & consistency ──
print(qs.stats.cagr(returns))          # compound annual growth
print(qs.stats.volatility(returns))    # annualised std
print(qs.stats.win_rate(returns))      # % of up days
print(qs.stats.avg_return(returns))    # mean daily return

# ── Full tear-sheet (one call!) ──
qs.reports.metrics(returns)            # table in terminal
qs.reports.html(returns,
    output="aapl_report.html")         # saves an HTML report
```
