```python
# range(stop) — 0 up to (not including) stop
for i in range(5):
    print(i)          # 0 1 2 3 4

# range(start, stop)
for i in range(1, 6):
    print(i)          # 1 2 3 4 5

# range(start, stop, step)
for i in range(0, 11, 2):
    print(i)          # 0 2 4 6 8 10
```

```python
import requests, time

TARGET = 500_000   # stop when BTC hits RM 500k

while True:
    r = requests.get("https://api.luno.com/api/1/ticker"
                      "?pair=XBTMYR")
    price = float(r.json()["last_trade"])
    print(f"BTC: RM {price:,.0f}")
    if price >= TARGET:
        break         # target reached!
    time.sleep(60)   # check every minute
```

```python
prices = {"apple": 4.50, "banana": 3.00}

# Loop the keys (the default)
for fruit in prices:
    print(fruit)

for fruit in prices:
    print(prices[fruit])

# Loop key + value together ✅
for fruit, price in prices.items():
    print(f"{fruit}: RM{price:.2f}")
```

```python
employees = [
    {"name": "Ahmad",    "monthly_salary": 2500, "company": "Petronas"},
    {"name": "Mei Ling", "monthly_salary": 5000, "company": "Maybank"},
    {"name": "Arif",     "monthly_salary": 8000, "company": "CIMB"},
]

# Loop through each employee
# Step 1: calculate annual salary (monthly × 12)
# Step 2: classify into B40 / M40 / T20
# Step 3: print the result

for employee in employees:
    annual_salary = employee["monthly_salary"]*12

    if annual_salary  > 80000:
        income_group = "T20"
    elif annual_salary >= 40000:
        income_group = "M40"
    else:
        income_group = "B40"

    print(f"{employee['name']} ({employee['company']}): {income_group}(RM {annual_salary}/year)")
```
